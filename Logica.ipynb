## Entrada, processamento e saída

### Variáveis e constantes:

Variáveis e constantes são utilizados para armazenar/assumir valores que serão utilizados no código.
* **Variáveis:** Todo dado que tem a possibilidade de ser alterado no decorrer do tempo;
* **Constante:** Quando um dado não tem nenhuma possibilidade de variar no decorrer do tempo.

### Tipos de dados:

* **Inteiros (int):** Valores positivos ou negativos, que não possuem uma parte fracionária (Ex.: 1, 30, 41, 12, -50).
* **Real (float):** Valores positivos ou negativos, que podem possuir uma parte fracionária (Ex.: 1.4, 6.7, 10.3, 100, -47).
* **Caracteres (char ou string - cadeia):** Qualquer elemento presente no teclado (ex.: 'Maria', 'João', 'M', 'F').
* **Lógico/boleano:** Verdadeiro ou falso (true/false).

### Declaração de variáveis:

* **Restrições para a criação de variáveis:**
    * Deve começar com letra ou _ ;
    * Só pode conter letras, números e _ ;
    * Não pode utilizar funções embutidas/bibliotecas.

* Para receber valores digitados pelo usuário via teclado, utiliza-se a função embutida ***input***.
    * **Atenção**: a função input sempre retornará um valor do tipo string.
* Caso o objetivo seja receber valores diferentes de texto/caracter, deve-se declarar o tipo de váriavel que a função retornará.

## Manipulação de strings

Funções utilizadas para a manipulação de string:
* `upper()`: torna todos os caracteres da string maiúsculos;
* `lower()`: torna todos os caracteres da string minúsculos;
* `capitalize()`: torna apenas o primeiro caracter da string maiúsculo;
* `replace()`: permite a substituição de caracteres na string;
* `find()`: permite a busca de caracteres na string.
* `len()`: retorna o tamanho da string.
* `strip()`: remove os espaços em branco no início/final da string.

Para percorrer uma string, segue-se o modelo: **nome_string[x:y]**
Sendo x o início e y o final.

* **Atenção**: O índice y não é incluído.

### Exemplos

```python
# Declaração e impressão da variável
a = 'casaco'
print(a)
```

```python
print(a.upper())
```

```python
print(a.lower())
```

```python
print(a.capitalize())
```

```python
print(a.replace('aco', 'inha'))
```

```python
print(a.find('aco')) # casaco
```

```python
print(len(a))
```

```python
metade_palavra = a[0:3]
print(metade_palavra)
```

```python
b = ' casaco '
print(b)
print(b.strip())
```

## Operadores condicionais

* Os operadores condicionais são utilizados para realizar escolhas dentro do algoritmo.
* Nos operadores condicionais são utilizados operadores lógicos e relacionais.
    * **Operadores lógicos:** and (tudo precisa ser verdadeiro), or (basta um ser verdadeiro) e not (inverte o resultado).
    * **Operadores relacionais:** > (maior), < (menor), >= (maior ou igual), <= (menor ou igual), == (igual) e != (diferente).
* No python, esse operador condicional se chama if/else (se/senão).


### Exemplos:

1. Leia a idade do usuário e classifique-o em:
* Criança – 0 a 12 anos
* Adolescente – 13 a 17 anos
* Adulto – acima de 18 anos
* Se o usuário digitar um número negativo, mostrar a mensagem que a idade é inválida

```python
idade = int(input('Digite a sua idade: '))

if 0 <= idade <= 12:
    print('Você é uma CRIANÇA com {} anos.'.format(idade))

elif 12 < idade <= 17:
    print('Você é um ADOLESCENTE com {} anos.'.format(idade))

elif idade >= 18:
    print('Você é um ADULTO com {} anos.'.format(idade))

else:
    print('A idade informada é inválida!')
    print('Olá, mundo!')

```

2. Desenvolva um algoritmo em Python para analisar as raízes de uma equação do segundo grau da forma: ax² + bx + c = 0

O usuário deve informar os valores de a, b e c.

Utilize os operadores condicionais para verificar se a equação possui raizes reais.


```python
import math
a = int(input('Informe o valor de "a" na equação ax² + bx + c = 0: '))
b = int(input('Informe o valor de "b" na equação ax² + bx + c = 0: '))
c = int(input('Informe o valor de "c" na equação ax² + bx + c = 0: '))

if a == 0:
    print('O valor de "a" deve ser diferente de zero.')

else:
    delta = b**2 - 4*a*c

    if delta < 0:
        print('A equação não admite raízes reais.')

    else:
        x_1 = (-b + math.sqrt(delta)) / (2*a)
        x_2 = (-b - math.sqrt(delta)) / (2*a)

        if x_1 == x_2:
            print('A equação admite uma única raiz igual a {}.'.format(x_1))

        else:
            print('A equação admite as seguintes raízes: {} e {}.'.format(x_1, x_2))
```

## Estruturas de repetição

* Quando temos uma sequência de comandos e desejamos que essa sequência seja repetida, utilizamos as estruturas de repetição.
* No python, os dois tipos de estruturas de repetição utilizados são:
    * **for**: no for, você precisa inicializar um contador e determinar o valor que ele será finalizado;
    * **while**: no while, enquanto uma condição for verdadeira, será realizado a sequência de códigos.
* Em estruturas de repetição, existem 2 tipos de variáveis bastante utilizadas:
    * **Contador**: sua função é contar quantas vezes o laço foi executado;
    * **Acumulador**: sua função é aculumar valores para futuras aplicações.

### Exemplos:

1. Ler 5 notas e informar a média. (utilizando for e while)

```python
soma = 0.0

for i in range(5): # [0, 1, 2, 3, 4]

    nota = float(input('Digite a {}º nota: '.format(i + 1)))

    while nota < 0 or nota > 10:
        print('Nota inválida! Digite uma nota entre 0 e 10.')
        nota = float(input('Digite novamente a {}º nota: '.format(i + 1)))

    soma += nota

media = soma / 5

print('A média aritmética é: {:.1f}'.format(media))
```

## Coleções

* Coleções são estruturas de dados que permitem armazenar vários valores dentro de uma única variável.
* Tipos de coleção:
    * **Tupla**: coleção ordenada e imutável (não pode alterar depois de criada).
    * **Lista**: coleção ordenada e mutável (pode alterar depois de criada).
    * **Dicionário**: coleção de pares chaves -> valor, não usa índice numérico e pode ser modificada.
    * **Conjuntos**: coleção de valores únicos, não permite elementos repetidos, não tem ordem fixa e é mutável.


### Exemplos:

```python
# Tupla
tupla = ('Clubes', 'Seleções', 'Ligas')
print(tupla)
```

```python
# Para mostras cada elemento da tupla, utiliza-se uma estrutura de repetição
for i in tupla:
    print(i)
```

```python
# Para acessar um elemento único da tupla, informa-se o índice
print(tupla[0])
```

```python
# Para retornar o índice sabendo o elemento, utiliza-se a função index
print(tupla.index('Seleções'))
```

```python
# Listas
l1 = ['Homo sapiens', 'Canis familiaris', 'Felis catus']
l2 = ['Xenopus laevis', 'Ailuropoda melanoleuca']
```

```python
# Concatenação de listas utilizando o operador (+)
l3 = l1 + l2
print(l3)
```

```python
# Repetição da lista utilziando o operador (*)
l2_2 = l2 * 2
print(l2_2)
```

```python
# Para acessar os elementos na lista, basta informar o índice
print(l1[0])
```

```python
# Para acessar uma sequência de valores, basta determinar o intervalo e passo
print(l2_2[:3])
```

```python
# Para adicionar valores a lista, utiliza-se a função append
l1.append('Gorila gorila')
print(l1)
```

```python
# Para remover valores da lista, utiliza-se a função remove
l1.remove('Felis catus')
print(l1)
```

```python
# Para acessar todos os elementos da lista, utiliza a estrutura de repetição
for elemento in l2_2:
    print(elemento)
```

```python
# Dicionários
coleta = {'Aedes aegypt': 32, 'Aedes albopictus': 22, 'Anopheles darlingi': 14}
```

```python
# Para acessar o valor de cada elemento, basta informar a chave
print(coleta['Aedes aegypt'])
```

```python
# Para modificar valores em cada elemento, informa-se a chave e utiliza-se o operador (=)
coleta['Anopheles darlingi'] = 11
print(coleta)
```

```python
# Para adicionar elementos, informa-se a chave e o valor correspondente
coleta['Rhodnius montenegrensis'] = 12
print(coleta)
```

```python
# Para deletar elementos, utiliza-se a função del
del coleta['Aedes aegypt']
print(coleta)
```

```python
# Para acessar todos os pares chaves/valores, utiliza-se a função items
print(coleta.items())
```

```python
# Para acessar apenas as chaves utiliza-se a função keys()
print(coleta.keys())
```

```python
# Para acessar apenas os valores utiliza-se a função values()
print(coleta.values())
```

```python
# Para atualizar os elementos do dicionário, utiliza-se a função update
coleta2 = {'Anopheles gambiae': 13, 'Anopheles deaneorum': 14}
coleta.update(coleta2)
print(coleta)
```

```python
# Para acessar todos os elementos de um dicionário, utiliza-se estruturas de repetição
for especie, num_especimes in coleta.items():
    print('Espécie: {}, número de espécimes coletados: {}'.format(especie, num_especimes))
```

```python
# Conjuntos
biomoleculas = ('proteína', 'ácidos nucleicos', 'carboidrato', 'lipídeo',
                'ácidos nucleicos', 'carboidrato', 'carboidrato', 'carboidrato')

# Dado uma tupla, utilizando a função set, transforma-se em um conjunto
conjunto = set(biomoleculas)
print(conjunto)
```

```python
# Utiliza-se a função intersection para retornar valores que se repetem em dois conjuntos
c1 = {1,2,3,4,5}
c2 = {3,4,5,6,7}
print(c1.intersection(c2))
```

```python
# Utiliza-se a função difference para mostrar elementos que existem em um conjunto mas não existem em outro
print(c1.difference(c2))
print(c2.difference(c1))
```
